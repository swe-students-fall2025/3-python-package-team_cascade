# 🐍 SsstudyPet Live Demo

Welcome to the **SsstudyPet** live demonstration! This notebook showcases all the core features of our gamified study tracking package.

## What is SsstudyPet?

SsstudyPet is a Python package that helps you:
- 📚 Track your study sessions
- ✅ Set and complete tasks
- 💰 Earn coins for completed work
- 🐍 Raise and care for a virtual ball python
- 📈 Monitor your progress and streaks

Let's dive into the features!

## 📦 Step 1: Import the Package

First, let's import all the functions we'll be using from SsstudyPet.

````python
# Import study session tracking functions
from study_pet.tracker import start_session, end_session, show_encouragement

# Import pet care functions
from study_pet.pet import feed_pet, rename_pet, set_morph, get_status, get_encouragement

# Import data management functions
from study_pet.data_manager import load_state, save_state

print("✅ All imports successful!")
````

## 🐍 Step 2: Check Initial Status

Let's see the starting state of our ball python!

````python
# Check the current status of our ball python
print(get_status())
````

## ✏️ Step 3: Personalize Your Pet

Let's give our ball python a name and choose a beautiful morph (color pattern)!

````python
# Rename our ball python
rename_pet("Demo")

# Set a beautiful morph (color pattern)
set_morph("Banana")

# Check the updated status
print("\n" + get_status())
````

## 💬 Step 4: Get Encouragement

Your ball python can motivate you with encouraging messages!

````python
# Get a few random encouragement messages
print("🌟 Here are some motivational messages from Demo:\n")
for i in range(3):
    print(get_encouragement())
````

## 📚 Step 5: Start a Study Session

Let's simulate starting a study session. We'll plan to complete 3 tasks.

**Note:** In a live demo, we'll use a mock input function to avoid interactive prompts.

````python
# Mock the input function to avoid interactive prompts during demo
import builtins
original_input = builtins.input

def mock_input(prompt):
    """Mock input function that returns pre-defined values"""
    if "How many tasks" in prompt:
        print(prompt + "3")  # Show what we're "typing"
        return "3"
    elif "did you complete" in prompt:
        print(prompt + "3")
        return "3"
    return ""

# Replace input with mock
builtins.input = mock_input

# Start the session
print("🚀 Starting study session...\n")
start_session()
````

## ⏱️ Step 6: Simulate Study Time

In a real scenario, you'd be studying now. Let's simulate some time passing.

````python
import time

# Simulate some study time
print("📖 Studying... (simulating 2 seconds)")
time.sleep(2)
print("✅ Study time complete!")
````

## 💪 Step 7: Get Encouragement During Session

While studying, you can get motivational messages anytime!

````python
# Get encouragement mid-session
show_encouragement()
````

## 🎉 Step 8: End the Study Session

Now let's end the session and claim our rewards!

````python
# Ensure mock input is still active
builtins.input = mock_input

# End the session
print("🏁 Ending study session...\n")
end_session()

# Restore original input function
builtins.input = original_input
````

## 📊 Step 9: Check Updated Status

Let's see how our progress looks after completing the session!

````python
# Check the updated status
print(get_status())

# Show the coins we earned
state = load_state()
print(f"\n💰 Current balance: {state['money']} coins")
````

## 🍽️ Step 10: Feed Your Ball Python

Now that we have coins, let's feed our python! Different foods have different costs and mood boosts.

````python
# Show available food options
print("🐍 Available Food Options:")
print("  - cricket (30 coins, +5 mood)")
print("  - mouse   (50 coins, +8 mood)")
print("  - rat     (80 coins, +15 mood)")
print("  - quail   (130 coins, +20 mood)")
print("  - rabbit  (150 coins, +25 mood)")
print()

# Feed the python a rat
feed_pet("rat")

# Check status after feeding
print("\n" + get_status())
````

## 📈 Step 11: View Detailed Statistics

Let's look at the raw data to see all the details!

````python
# Load and display all state data
state = load_state()

print("📊 Detailed Statistics:")
print("=" * 50)
print(f"🐍 Pet Name:              {state.get('name', 'Unnamed')}")
print(f"🎨 Morph:                 {state.get('morph', 'Normal')}")
print(f"⭐ Level:                 {state.get('level', 1)}")
print(f"✨ Experience:            {state.get('experience', 0):.1f} EXP")
print(f"💰 Money:                 {state.get('money', 0)} coins")
print(f"😊 Mood:                  {state.get('mood', 100)}/100")
print(f"⏱️  Total Study Time:     {state.get('total_study_time', 0):.2f} hours")
print(f"🔥 Streak:                {state.get('streak_days', 0)} days")
print(f"📅 Last Study:            {state.get('last_study_date', 'Never')}")
print(f"📝 Last Session Tasks:    {state.get('last_session_tasks_planned', 0)} planned")
print(f"✅ Tasks Completed:       {state.get('session_tasks_completed', 0)}")
print("=" * 50)
````

## 🎮 Step 12: Try Different Morphs

Ball pythons come in many beautiful morphs! Let's try a few different ones.

````python
# Show a few popular ball python morphs
print("🎨 Popular Ball Python Morphs:\n")

morphs_to_try = ["Albino", "Pastel", "Spider", "Clown"]

for morph in morphs_to_try:
    set_morph(morph)
    print()  # Add spacing

# Set it back to Banana for the final look
set_morph("Banana")
````

## 🎯 Key Features Summary

### Core Features:
1. **📚 Study Session Tracking** - Track start/end times with automatic time calculation
